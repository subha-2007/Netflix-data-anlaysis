```python
import pandas as pd
from google.colab import files

df = pd.read_csv('netflix_titles.csv')

df['duration_value'] = df['duration'].str.extract(r'(\d+)').astype('Int64')
df['duration_minutes'] = df.loc[df['type'] == 'Movie', 'duration_value']
df['duration_seasons'] = df.loc[df['type'] == 'TV Show', 'duration_value']
df.drop(columns=['duration_value'], inplace=True)

df.to_csv('netflix_titles_cleaned.csv', index=False)

files.download('netflix_titles_cleaned.csv')
```
